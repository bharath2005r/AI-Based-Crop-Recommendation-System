```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Optional: Better chart appearance
plt.style.use("ggplot")

# Load Dataset
df = pd.read_csv("../dataset/Crop_recommendation.csv")
```

```python
plt.figure(figsize=(12,6))

sns.countplot(x="label", data=df)

plt.xticks(rotation=90)

plt.title("Distribution of Crop Labels")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(df["N"], bins=20, kde=True)

plt.title("Distribution of Nitrogen")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(df["P"], bins=20, kde=True)

plt.title("Distribution of Phosphorus")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(df["K"], bins=20, kde=True)

plt.title("Distribution of Potassium")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(df["temperature"], bins=20, kde=True)

plt.title("Temperature Distribution")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(df["humidity"], bins=20, kde=True)

plt.title("Humidity Distribution")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(df["rainfall"], bins=20, kde=True)

plt.title("Rainfall Distribution")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(df["ph"], bins=20, kde=True)

plt.title("Soil pH Distribution")

plt.show()
```

```python
plt.figure(figsize=(10,8))

sns.heatmap(df.drop("label", axis=1).corr(),
            annot=True,
            cmap="coolwarm")

plt.title("Feature Correlation Heatmap")

plt.show()
```

```python
plt.figure(figsize=(14,6))

sns.boxplot(data=df.drop("label", axis=1))

plt.xticks(rotation=45)

plt.title("Boxplot of Numerical Features")

plt.show()
```
